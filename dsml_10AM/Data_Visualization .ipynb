```python
jpr_temp = np.random.randint(20, 40, 7).astype("float")
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
dehli_temp = np.random.randint(20, 40, 7).astype("float")
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

```python
fig = plt.figure(edgecolor = "red")
fig.set_facecolor("y")
ax = plt.gca()
ax.set_facecolor("cyan")
fig.set_linewidth(7)
plt.plot(days , dehli_temp , marker = "o" , mec = "r" , color = "black" , label = "dehli")
plt.plot(days , jpr_temp , marker = "o" , mec = "black" , color = "red" , label = "jaipur")
plt.grid(axis = "y")
plt.title("Temperature")
plt.xlabel("Day's" , color = "r")
plt.ylabel("Temp" , color = "b")
plt.tick_params("both" , direction = "inout" , length=9 , width = 5)
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(data["total_bill"] , color = "r" , marker = "o" , mec = "black" , linestyle = "")
plt.plot(data["day"])
plt.show()
```

```python
x = data.groupby(by = "day")["total_bill"].mean()
x
```

```python
plt.figure()
plt.plot(x , color = "g" , linestyle = "-.")
plt.xlabel("Average Total Bill")
plt.ylabel("Total bill Vs Day's")
plt.show()
```

```python
avg_bill = data.groupby(["day","sex"])["total_bill"].mean().unstack()
plt.figure()
plt.plot(avg_bill.index, avg_bill["Male"], marker="o", label="Male")
plt.plot(avg_bill.index, avg_bill["Female"], marker="o", label="Female")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.title("Average Total Bill Trend by Day and Sex")
plt.legend()
plt.show()
```

```python
q1 = np.quantile(df["total_bill"] , 0.25 )
q3 = np.quantile(df["total_bill"] , 0.75 )
print(q1 , q3 , sep = "\n")
```

```python
IQR = q3 - q1
print(IQR)
```

```python
lower = q1 - 1.5 * IQR 
upper = q3 + 1.5 * IQR
print(lower , upper , sep = "\n")
```
